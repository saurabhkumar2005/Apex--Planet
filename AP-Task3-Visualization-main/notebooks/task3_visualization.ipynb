```python
# ===============================
# Import Required Libraries
# ===============================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px

# Improve plot appearance
plt.style.use("ggplot")
sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python
# ===============================
# Load Dataset
# ===============================

df = pd.read_csv("../data/cleaned/ecommerce_dataset_cleaned.csv")

print("Dataset loaded successfully!")
```

```python
# ===============================
# Dataset Information
# ===============================

print("Shape of Dataset:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# ===============================
# Dataset Summary
# ===============================

print("Dataset Information")
print("-" * 50)

df.info()

print("\nStatistical Summary")
print("-" * 50)

display(df.describe())

print("\nMissing Values")
print("-" * 50)

display(df.isnull().sum())

print("\nDuplicate Rows")
print("-" * 50)

print(df.duplicated().sum())
```

```python
# ===============================
# Line Chart - Product ID vs Price
# ===============================

plt.figure(figsize=(12,5))

plt.plot(
    df["product_id"],
    df["price"],
    color="blue",
    linewidth=2
)

plt.title("Product ID vs Price")
plt.xlabel("Product ID")
plt.ylabel("Price")

plt.grid(True)

# Save Image
plt.savefig("../visualizations/png/line_chart_price.png", dpi=300)

plt.show()
```

```python
# ===============================
# Bar Chart - Category vs Total Units Sold
# ===============================

category_sales = df.groupby("category")["units_sold"].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))

category_sales.plot(
    kind="bar",
    color="skyblue",
    edgecolor="black"
)

plt.title("Total Units Sold by Category")
plt.xlabel("Category")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=45)

plt.tight_layout()

# Save Image
plt.savefig("../visualizations/png/bar_chart_category_sales.png", dpi=300)

plt.show()
```

```python
# ===============================
# Scatter Plot - Price vs Units Sold
# ===============================

plt.figure(figsize=(10,6))

plt.scatter(
    df["price"],
    df["units_sold"],
    color="green",
    alpha=0.7
)

plt.title("Price vs Units Sold")
plt.xlabel("Price")
plt.ylabel("Units Sold")

plt.grid(True)

# Save Image
plt.savefig("../visualizations/png/scatter_plot_price_units.png", dpi=300)

plt.show()
```

```python
# ===============================
# Histogram - Product Price
# ===============================

plt.figure(figsize=(10,6))

plt.hist(
    df["price"],
    bins=20,
    color="orange",
    edgecolor="black"
)

plt.title("Distribution of Product Price")
plt.xlabel("Price")
plt.ylabel("Frequency")

# Save Image
plt.savefig("../visualizations/png/histogram_price.png", dpi=300)

plt.show()
```

```python
# ===============================
# Heatmap - Correlation Matrix
# ===============================

plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include=["number"])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

# Save Image
plt.savefig("../visualizations/png/heatmap.png", dpi=300)

plt.show()
```

```python
# ===============================
# Pairplot
# ===============================

pairplot = sns.pairplot(
    df[["price", "units_sold", "rating"]],
    diag_kind="hist"
)

pairplot.fig.suptitle("Pairplot of Numerical Features", y=1.02)

# Save Image
pairplot.savefig("../visualizations/png/pairplot.png")

plt.show()
```

```python
# ===============================
# Boxen Plot - Category vs Price
# ===============================

plt.figure(figsize=(10,6))

sns.boxenplot(
    data=df,
    x="category",
    y="price",
    palette="Set2"
)

plt.title("Category vs Price (Boxen Plot)")
plt.xlabel("Category")
plt.ylabel("Price")

plt.xticks(rotation=45)

plt.tight_layout()

# Save Image
plt.savefig("../visualizations/png/boxen_plot.png", dpi=300)

plt.show()
```

```python
# ===============================
# Plotly Interactive Bar Chart
# ===============================

import plotly.express as px
import plotly.io as pio
import os

# VS Code Notebook ke liye browser renderer
pio.renderers.default = "browser"

# Category wise total units sold
category_sales = (
    df.groupby("category", as_index=False)["units_sold"]
      .sum()
)

# Create Interactive Bar Chart
fig = px.bar(
    category_sales,
    x="category",
    y="units_sold",
    color="category",
    title="Total Units Sold by Category",
    text_auto=True
)

# Improve Layout
fig.update_layout(
    xaxis_title="Category",
    yaxis_title="Total Units Sold",
    template="plotly_white",
    showlegend=False
)

# Create html folder if it doesn't exist
os.makedirs("../visualizations/html", exist_ok=True)

# Save Interactive HTML
fig.write_html("../visualizations/html/category_sales_interactive.html")

print("✅ Interactive HTML chart saved successfully!")

# Open chart in browser
fig.show()
```
