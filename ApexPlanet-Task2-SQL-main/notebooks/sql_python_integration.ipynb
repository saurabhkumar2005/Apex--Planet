```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("sqlite:///../database/ecommerce.db")
print("Database Connected Successfully!")
```

```python
query = "SELECT * FROM ecommerce LIMIT 5;"

df = pd.read_sql(query, engine)

df
```

# Business Question 1

## Top 5 Most Expensive Products

```python
query = """
SELECT product_name,
       category,
       price
FROM ecommerce
ORDER BY price DESC
LIMIT 5;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 2

## Category-wise Total Products

```python
query = """
SELECT category,
       COUNT(*) AS total_products
FROM ecommerce
GROUP BY category;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 3

## Average Price by Category

```python
query = """
SELECT category,
       AVG(price) AS average_price
FROM ecommerce
GROUP BY category;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 4

## Top 5 Highest Selling Products

```python
query = """
SELECT product_name,
       units_sold
FROM ecommerce
ORDER BY units_sold DESC
LIMIT 5;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 5

## Top 5 Highest Rated Products

```python
query = """
SELECT product_name,
       rating
FROM ecommerce
ORDER BY rating DESC
LIMIT 5;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 6

## Total Units Sold by Category

```python
query = """
SELECT category,
       SUM(units_sold) AS total_units
FROM ecommerce
GROUP BY category;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 7

## Products Out of Stock

```python
query = """
SELECT product_name,
       category
FROM ecommerce
WHERE in_stock = 'False';
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 8

## Products with Rating Greater Than 4

```python
query = """
SELECT product_name,
       rating
FROM ecommerce
WHERE rating > 4;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 9

## Highest Price by Category

```python
query = """
SELECT category,
       MAX(price) AS highest_price
FROM ecommerce
GROUP BY category;
"""

df = pd.read_sql(query, engine)

df
```

# Business Question 10

## Total Revenue by Category

```python
query = """
SELECT category,
       SUM(price * units_sold) AS total_revenue
FROM ecommerce
GROUP BY category;
"""

df = pd.read_sql(query, engine)

df
```
